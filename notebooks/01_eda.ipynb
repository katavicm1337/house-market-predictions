```python
missing = df.isnull().sum() # zbraja sve prazne celije za svaki stupac zasebno
missing = missing[missing > 0].sort_values(ascending=False)
missing
```

```python
postotak = (missing / len(df) * 100).round(2) # racuna koliki postotak cine nedostajuce vrijednosti za svaki stupac u odnosu na broj redova

missing_tablica = pd.DataFrame({
    "broj_praznih": missing,
    "postotak": postotak
})
missing_tablica # pretvara podatke u tablicu
```

```python
df.duplicated().sum() # provjera duplikata u datasetu
```

```python
# graf

plt.figure(figsize=(10, 6))
sns.barplot(x=missing_tablica["postotak"], y=missing_tablica.index, color="steelblue")
plt.xlabel("Postotak nedostajućih vrijednosti")
plt.ylabel("Stupac")
plt.title("Nedostajuće vrijednosti po stupcu")
plt.show()
```
